```python
from tensorflow.keras.applications.resnet_v2 import ResNet152V2, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
ResNet_model = ResNet152V2(weights='imagenet')
```

```python
img = image.load_img('/content/drive/My Drive/lab3/airplane.jpg', target_size=(224, 224))
```

```python
img
```

```python
x = image.img_to_array(img)
x = np.expand_dims(x, axis=0)
x = preprocess_input(x)
```

```python
x.shape
```

```python
preds = ResNet_model.predict(x)
```

```python
print('Результаты: ', decode_predictions(preds, top=3)[0])
```
